## Hindcast Analysis

```python
## Quinta 1995 (Only 5+ Observations) hindcast analysis

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

MODEL   = os.path.join('..', 'data', 'cosmos', 'quinta_hind_shoreline_prediction.csv')        # Datetime + T1..T10, daily (d/mm/yyyy)
OBS     = os.path.join('..', 'data', 'cosmos', 'quinta_shorelines_all.csv')      # Time + T1..T10, daily rows with blanks
OUT     = os.path.join('..', 'output', 'quinta_hindcast')
CAL_END = None                         # e.g. '2025-12-31' to mark the calibration/validation split

def load(path, dayfirst=False):
    d = pd.read_csv(path)
    d.index = pd.to_datetime(d[d.columns[0]], format='mixed', dayfirst=dayfirst)
    return d.drop(columns=d.columns[0]).sort_index()

mod = load(MODEL, dayfirst=True)
obs = load(OBS).dropna(how='all')
cols = [c for c in mod.columns if c in obs.columns]
SMOOTH = 30      # days; set to None for the raw daily line only

fig, axs = plt.subplots(len(cols), 1, figsize=(13, 1.7 * len(cols)), sharex=True, tight_layout=True)
for ax, c in zip(np.atleast_1d(axs), cols):
    ax.plot(mod.index, mod[c], '-', color='C3', lw=0.4, alpha=0.3)
    if SMOOTH:
        ax.plot(mod.index, mod[c].rolling(SMOOTH, center=True).mean(), '-', color='C3', lw=1.2, label='model')
    ax.plot(obs.index, obs[c], '.', color='0.3', ms=3, label='observed')
    ax.set_ylabel(f'{c} [m]')
    ax.grid(linestyle=':', color='0.6')
np.atleast_1d(axs)[0].legend(fontsize=8, ncol=2)
np.atleast_1d(axs)[-1].set_xlabel('year')
#fig.savefig(f'{OUT}_transects.png', dpi=200)
```

## Parameter and Skill Analysis

```python
"""Compare CoSMoS-COAST calibration runs: parameters and skill."""
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

PATH = os.path.join('..', 'data', 'cosmos')

RUNS = {                      # label -> (parameters csv, skill csv)
    'all95':  (os.path.join(PATH, 'quintaall_hind_parameters.csv'), os.path.join(PATH, 'quintaall_hind_skill_metrics.csv')),
    '2014_2': (os.path.join(PATH, 'quinta142_hind_parameters.csv'),    os.path.join(PATH, 'quinta142_hind_skill_metrics.csv')),
    '2014':   (os.path.join(PATH, 'quinta14_hind_parameters.csv'),  os.path.join(PATH, 'quinta14_hind_skill_metrics.csv')),
}
BOUNDS = {'DT_cs': (20, 300), 'DY_cs': (2, 100), 'DT_ls': (20, 365),
          'DY_ls': (-250, 250), 'c': (0.5, 1.5), 'vlt': (-1.5, 1), 'Y0': (-70, 70)}
OUT = 'run_comparison'
 
par = pd.concat({k: pd.read_csv(v[0], index_col=0) for k, v in RUNS.items()})
skl = pd.concat({k: pd.read_csv(v[1], index_col=0) for k, v in RUNS.items()})
par.index.names = skl.index.names = ['run', 'x']
transects = list(par.columns)
 
# ---------------------------------------------------------------- tables
def table(df, row):
    return df.xs(row, level='x').reindex(RUNS).round(2)
 
print('=' * 70, '\nSKILL (mean over transects)\n')
summary = pd.DataFrame({m: skl.xs(m, level='x').mean(axis=1) for m in skl.index.levels[1]}).reindex(RUNS)
print(summary.round(3).to_string())
 
for m in ['L', 'COR', 'RMSE','RMSEn','IDX']:
    print(f'\n{m} per transect\n' + table(skl, m).to_string())
 
print('\n' + '=' * 70, '\nPARAMETERS\n')
for p in par.index.levels[1]:
    t = table(par, p)
    lo, hi = BOUNDS.get(p, (None, None))
    if lo is not None:
        at = ((t <= lo + 0.01 * (hi - lo)) | (t >= hi - 0.01 * (hi - lo))).sum(axis=1)
        t = t.assign(**{'at bound': at})
    print(f'\n{p}  (bounds {lo} to {hi})\n' + t.to_string())
 
spread = pd.DataFrame({p: table(par, p).max() - table(par, p).min() for p in par.index.levels[1]}).T
print('\nspread across runs (max - min per transect)\n' + spread.round(2).to_string())
 
sign = {p: int((np.sign(table(par, p)).nunique() > 1).sum()) for p in ['DY_ls', 'vlt', 'Y0']}
print('\ntransects where the sign differs between runs:', sign)
 
# ---------------------------------------------------------------- figures
x = np.arange(len(transects))
pars = list(par.index.levels[1])
fig, axs = plt.subplots(len(pars), 1, figsize=(11, 2.0 * len(pars)), sharex=True, tight_layout=True)
for ax, p in zip(axs, pars):
    t = table(par, p)
    for k in RUNS:
        ax.plot(x, t.loc[k], 'o-', ms=4, lw=1, label=k)
    lo, hi = BOUNDS.get(p, (None, None))
    if lo is not None:
        ax.axhline(lo, color='0.6', ls='--', lw=0.8); ax.axhline(hi, color='0.6', ls='--', lw=0.8)
        pad = 0.08 * (hi - lo); ax.set_ylim(lo - pad, hi + pad)
    ax.set_ylabel(p); ax.grid(linestyle=':', color='0.7')
axs[0].legend(fontsize=8, ncol=len(RUNS))
axs[-1].set_xticks(x); axs[-1].set_xticklabels(transects); axs[-1].set_xlabel('transect')
#fig.savefig(f'{OUT}_parameters.png', dpi=200)
 
mets = ['RMSE', 'COR', 'STDn', 'L']
fig, axs = plt.subplots(len(mets), 1, figsize=(11, 2.0 * len(mets)), sharex=True, tight_layout=True)
for ax, m in zip(axs, mets):
    t = table(skl, m)
    for k in RUNS:
        ax.plot(x, t.loc[k], 'o-', ms=4, lw=1, label=k)
    if m in ('COR', 'STDn'):
        ax.axhline(1, color='0.6', ls='--', lw=0.8)
    ax.set_ylabel(m); ax.grid(linestyle=':', color='0.7')
axs[0].legend(fontsize=8, ncol=len(RUNS))
axs[-1].set_xticks(x); axs[-1].set_xticklabels(transects); axs[-1].set_xlabel('transect')
#fig.savefig(f'{OUT}_skill.png', dpi=200)
```

## EOF Analysis 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
 
MODEL    = os.path.join('..', 'data', 'cosmos', 'quinta14_hind_shoreline_prediction.csv')    # Datetime + T1..Tn, daily model output
OUT      = os.path.join('..', 'data', 'cosmos', 'output', 'quinta_eof')
DEMEAN   = True      # False = Miller & Dean (mode 1 = "mean shoreline"); True = Jakob's covariance version
N_SHOW   = 4         # modes to plot (Miller & Dean report the first four)
RESAMPLE = None      # e.g. '30D' for PC spectra, as Miller & Dean do before spectral analysis
SPACING  = None      # optional alongshore distance per transect [m]; None = transect index
TRANSECTS = ['T3', 'T4', 'T5', 'T6', 'T7', 'T8']   # section B only; None = all
PERIOD    = (None, None)     #'2014-01-01'              # calibration period; (None, None) = full hindcast
```

```python
# ------------------------------------------------------------------ 0. data
mod = pd.read_csv(MODEL)
mod.index = pd.to_datetime(mod[mod.columns[0]], format='mixed', dayfirst=True)
mod = mod.drop(columns=mod.columns[0]).sort_index().dropna(how='any')
if RESAMPLE:
    mod = mod.resample(RESAMPLE).mean().dropna(how='any')
if TRANSECTS:
    mod = mod[TRANSECTS]
if PERIOD[0] or PERIOD[1]:
    mod = mod.loc[PERIOD[0]:PERIOD[1]]
names = list(mod.columns)
x = np.arange(len(names)) if SPACING is None else np.asarray(SPACING, float)
```

```python
# ------------------------------------------------------------------ 1. pre-EOF statistics
stats = pd.DataFrame({'mean': mod.mean(), 'std': mod.std(),
                      'range': mod.max() - mod.min()}).round(2)
print('pre-EOF statistics (Miller & Dean section 2)\n' + stats.to_string())
```

```python
# ------------------------------------------------------------------ 2. EOF decomposition
Y = mod.to_numpy().T                      # [nx, nt] : space x time
nx, nt = Y.shape
mean_shoreline = Y.mean(axis=1)
if DEMEAN:
    Y = Y - mean_shoreline[:, None]
 
U, S, Vt = np.linalg.svd(Y, full_matrices=False)   # equivalent to eig of A = YY'/(nx nt)
lam = S ** 2 / (nx * nt)                           # eigenvalues
p   = 100 * lam / lam.sum()                        # Eq. 12: percent variance per mode
a   = np.sqrt(lam * nx * nt)                       # normalising factor a_k (= S)
e   = U                                            # spatial eigenfunctions e_k(x), orthonormal
c   = Vt                                           # temporal eigenfunctions c_k(t)
 
# sign convention (not in the paper): make each mode's largest-|e| weight positive
for k in range(len(S)):
    if e[np.argmax(np.abs(e[:, k])), k] < 0:
        e[:, k] *= -1
        c[k, :] *= -1
 
eofs = pd.DataFrame(e[:, :N_SHOW], index=names, columns=[f'EOF{k+1}' for k in range(N_SHOW)])
pcs  = pd.DataFrame((a[:N_SHOW, None] * c[:N_SHOW]).T, index=mod.index,
                    columns=[f'PC{k+1}' for k in range(N_SHOW)])   # a_k c_k(t), in metres
var  = pd.Series(p[:N_SHOW], index=eofs.columns, name='variance_%').round(2)
 
print('\nvariance explained (%):\n' + var.to_string())
print('\nspatial eigenfunctions e_k(x):\n' + eofs.round(3).to_string())
print(f'\nfirst {N_SHOW} modes explain {p[:N_SHOW].sum():.1f}% of the variability')
#eofs.to_csv(f'{OUT}_eofs.csv'); pcs.to_csv(f'{OUT}_pcs.csv'); var.to_csv(f'{OUT}_variance.csv')
 
# nodes: sign changes along the beach (Miller & Dean interpretation rules)
for k in range(N_SHOW):
    s = np.sign(e[:, k])
    nodes = [f'{names[i]}-{names[i+1]}' for i in range(nx - 1) if s[i] != s[i + 1]]
    kind = 'no node: in-phase (cross-shore)' if not nodes else f'node(s) between {", ".join(nodes)}'
    print(f'EOF{k+1} ({p[k]:5.1f}%): {kind}')
 
```

```python
# ------------------------------------------------------------------ 3. figures
fig, axs = plt.subplots(N_SHOW, 2, figsize=(13, 2.2 * N_SHOW), tight_layout=True,
                        gridspec_kw={'width_ratios': [1, 3]})
for k in range(N_SHOW):
    axs[k, 0].plot(x, e[:, k], 'o-', color='C0')
    axs[k, 0].axhline(0, color='k', lw=0.8)
    axs[k, 0].set_ylabel(f'EOF{k+1}\n{p[k]:.1f}%')
    axs[k, 0].set_xticks(x); axs[k, 0].set_xticklabels(names, fontsize=7)
    axs[k, 1].plot(pcs.index, pcs.iloc[:, k], '-', color='C3', lw=0.6)
    axs[k, 1].axhline(0, color='k', lw=0.8); axs[k, 1].set_ylabel('a$_k$c$_k$(t) [m]')
    for a_ in (axs[k, 0], axs[k, 1]):
        a_.grid(linestyle=':', color='0.7')
axs[-1, 0].set_xlabel('transect'); axs[-1, 1].set_xlabel('year')
#fig.savefig(f'{OUT}_modes.png', dpi=200)
 
# combined eigenfunctions y_k(x,t) = a_k c_k(t) e_k(x)   (Miller & Dean Eq. 13)
fig, axs = plt.subplots(1, N_SHOW, figsize=(3.2 * N_SHOW, 5), sharey=True, tight_layout=True)
for k in range(N_SHOW):
    yk = np.outer(e[:, k], a[k] * c[k])
    v = np.abs(yk).max()
    im = axs[k].pcolormesh(x, mod.index, yk.T, cmap='RdBu_r', vmin=-v, vmax=v, shading='nearest')
    axs[k].set_title(f'mode {k+1} ({p[k]:.1f}%)'); axs[k].set_xticks(x)
    axs[k].set_xticklabels(names, fontsize=7, rotation=90)
    plt.colorbar(im, ax=axs[k], label='m' if k == N_SHOW - 1 else None)
axs[0].set_ylabel('year')
#fig.savefig(f'{OUT}_combined.png', dpi=200)
 
# reconstruction with L modes (Eq. 13) vs the full signal
L = N_SHOW
recon = (e[:, :L] * a[:L]) @ c[:L] + (0 if DEMEAN is False else mean_shoreline[:, None])
err = np.sqrt(np.mean((recon - mod.to_numpy().T) ** 2))
print(f'\nreconstruction with {L} modes: RMS difference from the full hindcast = {err:.2f} m')
```
